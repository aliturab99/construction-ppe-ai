```python
import cv2
import torch
import time
from ultralytics import YOLO
```

```python
# model = YOLO("./runs/detect/run/ppe/first_experiment/weights/best.pt") # this model is generated by yolo26n
model = YOLO("./runs/detect/run/ppe/construction_ppe/weights/best.pt") # this model is generated by yolo26m
# model = YOLO("./yolo26m.pt")
```

```python
vedio = cv2.VideoCapture(0)

if not vedio.isOpened():
    raise RuntimeError("Could Not Open Vedio")

fps = vedio.get(cv2.CAP_PROP_FPS)
width = vedio.get(cv2.CAP_PROP_FRAME_WIDTH)
height = vedio.get(cv2.CAP_PROP_FRAME_HEIGHT)

print(fps)
print(width)
print(height)

previous_time = time.time()

while vedio.isOpened():
    ret, frame = vedio.read()

    if not ret:
        break

    result = model.predict(
        frame,
        conf= 0.35,
        device=0,
    )

    annotated_frame = result[0].plot()

    current_time = time.time()
    processing_fps = 1 / max(current_time - previous_time, 0.0001)
    previous_time = current_time

    cv2.putText(
        annotated_frame,
        f"FPS: {processing_fps}",
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0,255,0),
        2
    )

    cv2.imshow("Construction PPE - Yolo detection", annotated_frame)

    if cv2.waitKey(int(1000 / fps)) & 0xff == ord("q"):
        break

vedio.release()
cv2.destroyAllWindows()
```
